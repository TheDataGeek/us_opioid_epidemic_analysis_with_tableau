### CREATE A FUNCTION TO CLEAN DATA

```python
# define function to clean data
def load_cdc_csv(filepath: str) -> pd.DataFrame:

    # load csv as string
    df = pd.read_csv(filepath, dtype = str)

    # CDC WONDER appends metadata below data rows
    # valid data rows have NaN in 'Notes' column
    # return index of first row with metadata
    cutoff = df[df['Notes'].notna()].index[0]

    # drop rows with metadata
    df_clean = df.iloc[:cutoff]

    # drop 'Notes' column
    df_clean = df_clean.drop(columns = 'Notes')
    
    return df_clean
```

### LEVERAGE FUNCTION TO LOAD & CLEAN MULTIPLE DATA FILES
### CONCATENATE AND SAVE CLEANED DATA TO PROCESSED FOLDER

```python
# load csv files
files = [
    '../data/raw/multiple_cause_of_death_1999_2020_T400_T401.csv',
    '../data/raw/multiple_cause_of_death_1999_2020_T402.csv',
    '../data/raw/multiple_cause_of_death_1999_2020_T403_T404.csv',
    '../data/raw/multiple_cause_of_death_1999_2020_T405_T406.csv'
]

# load process csv files into list
df_combine = [load_cdc_csv(f) for f in files]

# concatenate
df_combine = pd.concat(df_combine, ignore_index=True)

print(f"Shape: {df_combine.shape}")
print(f"Unique ICD codes: {df_combine['Multiple Cause of death Code'].unique()}")
print(f"Year range: {df_combine['Year'].min()} – {df_combine['Year'].max()}")

# write to processed directory
df_combine.to_csv(
    '../data/processed/multiple_cause_of_death_1999_2020_processed.csv', 
    index = False
)

print('Saved: ../data/processed/multiple_cause_of_death_1999_2020_processed.csv')
```
