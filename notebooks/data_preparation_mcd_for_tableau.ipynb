### Convert `Crude Rate` to numerical values. Convert non-numerical values to `NaN`

```python
df_mcd['Crude Rate'] = pd.to_numeric(df_mcd['Crude Rate'], errors = 'coerce')
df_mcd['Crude Rate'].dtype
```

#### Return count of `NaN` values

```python
df_mcd['Crude Rate'].isna().sum()
```

### Return % of `NaN` values in `Crude Rate`

```python
crude_rate_null_pct = (df_mcd['Crude Rate'].isna().sum() / df_mcd.shape[0] * 100).round(2)
print(f" Percent of null values in Crude Rate column: {crude_rate_null_pct}")
```

**Important:** Crude Rate = Count / Population * 100,000. Rates are marked as "unreliable" when the death count is less than 20. Reference: https://wonder.cdc.gov/wonder/help/mcd.html

### `Crude Rate` is unusable given that has ~54% `NaN` values

### Create custom `Crude Rate Calculated` column

```python
df_mcd['Crude Rate Calculated'] = (df_mcd['Deaths'] / df_mcd['Population'] * 100000).round(2)
```

### Confirm `Crude Rate Calculated`

### Columns we can safely drop `Year Code`, `Five-Year Age Groups Code`, `Sex Code`, `Crude Rate` - keeping `State Code` in case Tableau can read it

```python
df_mcd_clean = df_mcd.drop(
    columns = [
        'Year Code', 
        'Five-Year Age Groups Code', 
        'Sex Code', 
        'Crude Rate'
    ]
)
```
