## 4. Create reusable function to remove metadata and format

```python
def remove_metadata(filepath: str) -> pd.DataFrame:
    
    # load file
    df = pd.read_csv(filepath, dtype=str)
    
    # assign index of first row with metadata to cutoff
    cutoff = df[df['Notes'].notna()].index[0]
    
    # drop rows with metadata
    df_sans_metadata = df.iloc[:cutoff]
    
    # drop 'Notes' column
    df_sans_metadata = df_sans_metadata.drop(columns='Notes')

    # capitalize 'Multiple Cause of death' to 'Multiple Cause of Death'
    df_sans_metadata.rename(columns = {
        'Multiple Cause of death':'Multiple Cause of Death',
        'Multiple Cause of death Code':'Multiple Cause of Death Code'
    }, inplace = True)

    return df_sans_metadata

# load and concatenate all files
files = [f'../data/raw/YEAR_STATE_AGE_MCD/YEAR_STATE_AGE_MCD_{year}.csv' for year in range(1999, 2021)]
df_clean = pd.concat([remove_metadata(f) for f in files], ignore_index=True)

print(f"Shape: {df_clean.shape}")
print(f"Unique ICD codes: {df_clean['Multiple Cause of Death Code'].unique()}")
print(f"Year range: {df_clean['Year'].min()} – {df_clean['Year'].max()}")

# write to processed directory
df_clean.to_csv(
    '../data/processed/YEAR_STATE_AGE_MCD_1999_2020_preprocessed.csv', 
    index = False
)

print('Saved: ../data/processed/YEAR_STATE_AGE_MCD_1999_2020_preprocessed.csv')
```
