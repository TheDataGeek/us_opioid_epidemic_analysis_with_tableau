```python
print(f"Shape: {df_ucd_clean.shape}")
print(f"Unique ICD codes: {df_ucd_clean['Multiple Cause of death Code'].unique()}")
print(f"Year range: {df_ucd_clean['Year'].min()} – {df_ucd_clean['Year'].max()}")

# write to processed directory
df_ucd_clean.to_csv(
    '../data/processed/underlying_cause_of_death_1999_2020_clean.csv', 
    index = False
)

print('Saved: ../data/processed/underlying_cause_of_death_1999_2020_clean.csv')
```
