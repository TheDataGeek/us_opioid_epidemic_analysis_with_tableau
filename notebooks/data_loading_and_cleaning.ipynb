```python
# define function to clean data
def load_cdc_csv(filepath: str) -> pd.DataFrame:

    # load csv as string
    df = pd.read_csv(filepath, dtype = str)

    # CDC WONDER appends metadata below data rows
    # valid data rows have NaN in 'Notes' column
    # return index of first row with metadata
    cutoff = df[df['Notes'].notna()].index[0]

    # drop rows with metadata
    df = df.iloc[:cutoff]

    # drop 'Notes' column
    df = df.drop(columns = 'Notes')
    
    return df
```

```python
# load csv files
files = [
    '../data/raw/multiple_cause_of_death_1999_2000_T400_T401.csv',
    '../data/raw/multiple_cause_of_death_1999_2000_T402.csv',
    '../data/raw/multiple_cause_of_death_1999_2000_T403_T404.csv',
    '../data/raw/multiple_cause_of_death_1999_2000_T405_T406.csv'
]

# load process csv files into list
df_complete = [load_cdc_csv(f) for f in files]

# concatenate
df_combined = pd.concat(df_complete, ignore_index=True)

print(f"Shape: {df_combined.shape}")
print(f"Unique ICD codes: {df_combined['Multiple Cause of death Code'].unique()}")
print(f"Year range: {df_combined['Year'].min()} – {df_combined['Year'].max()}")

# write to processed directory
df_combined.to_csv(
    '../data/processed/multiple_cause_of_death_1999_2000_T400_T406.csv', 
    index = False
)

print('Saved: ../data/processed/multiple_cause_of_death_1999_2000_T400_T406.csv')
```
