## 4. Create reusable function to remove metadata

```python
def remove_metadata(filepath: str) -> pd.DataFrame:
    
    # load file
    df = pd.read_csv(filepath, dtype=str)
    
    # assign index of first row with metadata to cutoff
    cutoff = df[df['Notes'] == '---'].index[0]
    
    # drop rows with metadata
    df_sans_metadata = df.iloc[:cutoff]

    # drop rows where 'Notes' == 'Total'
    df_sans_metadata = df_sans_metadata[df_sans_metadata['Notes'] != 'Total']
    
    # drop 'Notes' column
    df_sans_metadata = df_sans_metadata.drop(columns='Notes')
    
    return df_sans_metadata

# load single file
filepath = '../data/raw/YEAR_GENDER_AGE_1999_2020.csv'
df_clean = remove_metadata(filepath)

print(f"Shape: {df_clean.shape}")
print(f"Year range: {df_clean['Year'].min()} – {df_clean['Year'].max()}")

# write to processed directory
df_clean.to_csv(
    '../data/processed/YEAR_GENDER_AGE_1999_2020_preprocessed.csv', 
    index = False
)

print('Saved: ../data/processed/YEAR_GENDER_AGE_1999_2020_preprocessed.csv')
```
